```python
# reusable plotting function
def plot_optimizer_comparison(data, learning_rate, metric='overall_accuracy', figsize=(10, 6)):
    """
    Plot comparison between AdamW and Muon optimizers for a given learning rate.
    
    Args:
        data: Dictionary containing optimizer data
        learning_rate: String representing the learning rate (e.g., '1e-6')
        metric: Which metric to plot (default: 'overall_accuracy')
        figsize: Figure size tuple
    """
    plt.figure(figsize=figsize)
    
    # plot adamw data
    if 'AdamW' in data and learning_rate in data['AdamW']:
        adamw_data = data['AdamW'][learning_rate]
        plt.plot(adamw_data['samples'], adamw_data[metric], 
                'o-', label=f'AdamW LR:{learning_rate}', 
                linewidth=2, markersize=8)
    
    # plot muon data
    if 'Muon' in data and learning_rate in data['Muon']:
        muon_data = data['Muon'][learning_rate]
        plt.plot(muon_data['samples'], muon_data[metric], 
                's-', label=f'Muon LR:{learning_rate}', 
                linewidth=2, markersize=8)
    
    # format the plot
    plt.xlabel('Number of Samples', fontsize=12)
    plt.ylabel(metric.replace('_', ' ').title(), fontsize=12)
    plt.title(f'OSFT on QuaLITY (RR=0.5): {metric.replace("_", " ").title()} vs Training Samples (LR: {learning_rate})', 
              fontsize=14)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.ticklabel_format(style='plain', axis='x')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# example usage - recreate the 1e-6 plot
plot_optimizer_comparison(data, '1e-6', metric='overall_accuracy')
```

```python
# function to plot all learning rates for both optimizers
def plot_all_learning_rates(data, metric='overall_accuracy', figsize=(12, 8)):
    """
    Plot all learning rates for both optimizers on a single plot.
    """
    plt.figure(figsize=figsize)
    
    # define color gradients for each optimizer
    # lower learning rates get lighter colors, higher get darker
    lr_order = ['1e-6', '5e-6', '2e-5', '1e-4']  # from lowest to highest
    
    # create color gradients: lighter to darker
    adamw_colors = {
        '1e-6': '#ffb3b3',  # light red
        '5e-6': '#ff6666',  # medium-light red
        '2e-5': '#ff1a1a',  # medium-dark red
        '1e-4': '#cc0000'   # dark red
    }
    
    muon_colors = {
        '1e-6': '#b3d9ff',  # light blue
        '5e-6': '#66b3ff',  # medium-light blue
        '2e-5': '#1a8cff',  # medium-dark blue
        '1e-4': '#0066cc'   # dark blue
    }
    
    color_maps = {'AdamW': adamw_colors, 'Muon': muon_colors}
    markers = {'AdamW': 'o', 'Muon': 's'}
    
    for optimizer in ['AdamW', 'Muon']:
        if optimizer in data:
            for lr in lr_order:  # plot in order from lowest to highest
                if lr in data[optimizer]:
                    lr_data = data[optimizer][lr]
                    plt.plot(lr_data['samples'], lr_data[metric], 
                            marker=markers[optimizer], color=color_maps[optimizer][lr],
                            label=f'{optimizer} LR:{lr}', 
                            linewidth=2, markersize=6, alpha=0.9)
    
    plt.xlabel('Number of Samples', fontsize=12)
    plt.ylabel(metric.replace('_', ' ').title(), fontsize=12)
    plt.title(f'OSFT on QuaLITY (RR=0.5): {metric.replace("_", " ").title()} vs Training Samples', 
              fontsize=14)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.ticklabel_format(style='plain', axis='x')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# plot all learning rates
plot_all_learning_rates(data, metric='overall_accuracy')
```

```python
# function to create subplot grid for each learning rate
def plot_lr_comparison_grid(data, metric='overall_accuracy', figsize=(15, 10)):
    """
    Create a subplot grid showing each learning rate comparison separately.
    """
    learning_rates = ['1e-6', '5e-6', '2e-5', '1e-4']
    fig, axes = plt.subplots(2, 2, figsize=figsize)
    axes = axes.flatten()
    
    for idx, lr in enumerate(learning_rates):
        ax = axes[idx]
        
        # plot adamw data
        if 'AdamW' in data and lr in data['AdamW']:
            adamw_data = data['AdamW'][lr]
            ax.plot(adamw_data['samples'], adamw_data[metric], 
                    'o-', label=f'AdamW', color='blue',
                    linewidth=2, markersize=6)
        
        # plot muon data
        if 'Muon' in data and lr in data['Muon']:
            muon_data = data['Muon'][lr]
            ax.plot(muon_data['samples'], muon_data[metric], 
                    's-', label=f'Muon', color='orange',
                    linewidth=2, markersize=6)
        
        ax.set_xlabel('Number of Samples', fontsize=10)
        ax.set_ylabel(metric.replace('_', ' ').title(), fontsize=10)
        ax.set_title(f'Learning Rate: {lr}', fontsize=12)
        ax.legend(fontsize=10)
        ax.grid(True, alpha=0.3)
        ax.tick_params(axis='x', rotation=45)
    
    plt.suptitle(f'OSFT on QuaLITY (RR=0.5): {metric.replace("_", " ").title()} Comparison', 
                 fontsize=16)
    plt.tight_layout()
    plt.show()

# create the comparison grid
plot_lr_comparison_grid(data, metric='overall_accuracy')
```
