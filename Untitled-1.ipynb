```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Extract data from the evaluation results
muon_data = {
    5e-6: [0.4675, 0.4723, 0.4819, 0.4627, 0.4723],
    1e-5: [0.5036, 0.4892, 0.494, 0.5036, 0.5036],
    2e-5: [0.5084, 0.4988, 0.5012, 0.5012, 0.5084],
    5e-5: [0.1952, 0.1952, 0.2193, 0.188, 0.1952],
    1e-4: [0.0024, 0.0024, 0.0024, 0.0024, 0.0024]
}

adamw_data = {
    1e-6: [0.4892, 0.4675, 0.4795, 0.4795, 0.4795],
    2e-6: [0.4892, 0.506, 0.506, 0.5012, 0.4964],
    5e-6: [0.5253, 0.5253, 0.5229, 0.5253, 0.5253],
    1e-5: [0.506, 0.506, 0.506, 0.5181, 0.4988],
    2e-5: [0.4313, 0.4361, 0.4361, 0.4361, 0.4361]
}

def calculate_stats(data_dict):
    """Calculate mean and 95% confidence interval for each learning rate"""
    learning_rates = []
    means = []
    ci_lower = []
    ci_upper = []
    
    for lr, scores in sorted(data_dict.items()):
        scores_array = np.array(scores)
        mean_score = np.mean(scores_array)
        sem = stats.sem(scores_array)  # standard error of the mean
        ci = stats.t.interval(0.95, len(scores_array)-1, loc=mean_score, scale=sem)
        
        learning_rates.append(lr)
        means.append(mean_score)
        ci_lower.append(ci[0])
        ci_upper.append(ci[1])
    
    return learning_rates, means, ci_lower, ci_upper

# Calculate statistics for both optimizers
muon_lr, muon_means, muon_ci_lower, muon_ci_upper = calculate_stats(muon_data)
adamw_lr, adamw_means, adamw_ci_lower, adamw_ci_upper = calculate_stats(adamw_data)

# Create the plot
plt.figure(figsize=(12, 8))

# Plot Muon results
plt.errorbar(muon_lr, muon_means, 
             yerr=[np.array(muon_means) - np.array(muon_ci_lower), 
                   np.array(muon_ci_upper) - np.array(muon_means)],
             fmt='b-o', label='Muon', color='blue', linewidth=2, markersize=8, capsize=5)

# Plot AdamW results
plt.errorbar(adamw_lr, adamw_means,
             yerr=[np.array(adamw_means) - np.array(adamw_ci_lower),
                   np.array(adamw_ci_upper) - np.array(adamw_means)],
             fmt='r-o', label='AdamW', color='red', linewidth=2, markersize=8, capsize=5)

plt.xlabel('Learning Rate', fontsize=12)
plt.ylabel('Overall Accuracy', fontsize=12)
plt.title('Overall Accuracy vs Learning Rate: Muon vs AdamW\n(with 95% Confidence Intervals)', fontsize=14)
plt.xscale('log')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.tight_layout()

# Add some styling
plt.ylim(0, 0.6)
plt.show()

# Print summary statistics
print("Muon Results:")
for i, lr in enumerate(muon_lr):
    print(f"  LR {lr:.0e}: {muon_means[i]:.4f} (95% CI: [{muon_ci_lower[i]:.4f}, {muon_ci_upper[i]:.4f}])")

print("\nAdamW Results:")
for i, lr in enumerate(adamw_lr):
    print(f"  LR {lr:.0e}: {adamw_means[i]:.4f} (95% CI: [{adamw_ci_lower[i]:.4f}, {adamw_ci_upper[i]:.4f}])")
```
